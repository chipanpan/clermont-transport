```python
import json
import pandas as pd
from ingest_t2c_to_gcs import fetch_and_parse_gtfs_rt

# 1. Fetch live parsed records from T2C endpoint
records = fetch_and_parse_gtfs_rt()

# 2. Convert records to a DataFrame
df = pd.DataFrame(records)

# 3. Data shape
df.shape
```

```python
# Convert Epoch integers to human-readable datetime (Europe/Paris timezone)
df['arrival_datetime'] = pd.to_datetime(df['arrival_time'], unit='s', utc=True).dt.tz_convert('Europe/Paris')
df['departure_datetime'] = pd.to_datetime(df['departure_time'], unit='s', utc=True).dt.tz_convert('Europe/Paris')

# View active routes and stop delays
df[['route_id', 'trip_id', 'stop_id', 'stop_sequence', 'delay_seconds', 'arrival_datetime']].head(10)
```
